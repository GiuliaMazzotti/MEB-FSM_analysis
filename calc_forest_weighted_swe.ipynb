```python
import xarray as xr
import numpy as np


path_cro=r'data\Cro_opn.nc'
path_meb=r'data\MEBCro_for.nc'
path_handmade =r'data\MEBCro_all.nc'
path_fsm = r'data\FSM_all.nc'

ds_cro = xr.open_dataset(path_cro,decode_times=True)
ds_meb = xr.open_dataset(path_meb,decode_times=True)
ds_all = xr.open_dataset(path_handmade,decode_times=True)
ds_fsm = xr.open_dataset(path_fsm,decode_times=True)

ds_all["swet_all_corr"] = ds_meb["swet_all"]*ds_fsm["forest"]+ds_cro["swet_all"]*(1-ds_fsm["forest"])
ds_cro
```

```python
ds_all.to_netcdf("data/MEBCro_all_corr.nc")
```
